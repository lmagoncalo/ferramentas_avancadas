```python
# 1. Imports
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 2. Tiny dataset
texts = [
    "I love this product",
    "This is amazing",
    "I hate this",
    "This is terrible",
    "Absolutely fantastic experience",
    "Worst thing ever"
]

labels = [1, 1, 0, 0, 1, 0]  # 1 = positive, 0 = negative

# 3. Convert text → word counts (Bag of Words)
vectorizer = CountVectorizer()
data = vectorizer.fit_transform(texts)

print(vectorizer.vocabulary_)

# 4. Train model
model = MultinomialNB()
model.fit(data, labels)

# 5. Test sentences
test_sentences = ["I love it", "This is the worst"]
X_test = vectorizer.transform(test_sentences)
predictions = model.predict(X_test)

# 6. Show results
for text, pred in zip(test_sentences, predictions):
    print(f"{text} → {'Positive' if pred == 1 else 'Negative'}")
```

```python
from collections import Counter
from collections import defaultdict

text = "this is a simple example text to show byte pair encoding in action"
text = list(text)

print("Original text:", text)

alphabet = []

for letter in text:
    if letter not in alphabet:
        alphabet.append(letter)
alphabet.sort()

vocab = ["<|endoftext|>"] + alphabet.copy()
print("Initial vocab:", vocab)

vocab_size = 35

while len(vocab) < vocab_size:
    pair_freqs = defaultdict(int)
    for i in range(len(text) - 1):
        pair = (text[i], text[i + 1])
        pair_freqs[pair] += 1

    pair_freqs = sorted(pair_freqs.items(), key=lambda kv: kv[1], reverse=True)
    most_freq_pair = pair_freqs[0]

    # The found pair only appears once
    if most_freq_pair[1] == 1:
        break

    vocab.append(most_freq_pair[0][0] + most_freq_pair[0][1])

    print("Most frequent pair:", most_freq_pair)
    replaced = []
    i = 0
    while i < len(text):
        if i < len(text) - 1 and (text[i], text[i + 1]) == most_freq_pair[0]:
            replaced.append(text[i] + text[i + 1])
            i += 2  # skip both characters
        else:
            replaced.append(text[i])
            i += 1

    text = replaced

print(vocab)
```
