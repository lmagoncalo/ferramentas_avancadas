```python
from sklearn.datasets import fetch_california_housing, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
# 2. Diabetes Dataset (smaller regression dataset for comparison)
dataset = fetch_california_housing()
data, target = dataset.data, dataset.target

# Split Diabetes data into training and test sets
X_train, X_test, y_train, y_test = train_test_split(
    data, target, test_size=0.2, random_state=42
)

model = GradientBoostingRegressor(random_state=42)

model.fit(X_train, y_train)

y_pred_train = model.predict(X_train)
y_pred_test = model.predict(X_test)

rf_train_r2 = r2_score(y_train, y_pred_train)
rf_test_r2 = r2_score(y_test, y_pred_test)
rf_train_mse = mean_squared_error(y_train, y_pred_train)
rf_test_mse = mean_squared_error(y_test, y_pred_test)
rf_train_mae = mean_absolute_error(y_train, y_pred_train)
rf_test_mae = mean_absolute_error(y_test, y_pred_test)

print(f"R² on training set  : {rf_train_r2:.4f}")
print(f"R² on test set      : {rf_test_r2:.4f}")
print(f"MSE on training set : {rf_train_mse:.4f}")
print(f"MSE on test set     : {rf_test_mse:.4f}")
print(f"MAE on training set : {rf_train_mae:.4f}")
print(f"MAE on test set     : {rf_test_mae:.4f}")
```

```python
# Define model
model = GradientBoostingRegressor(random_state=42)

# Keep grid small for speed
param_grid = {
    "n_estimators": [50, 100, 200],
    "learning_rate": [0.01, 0.05, 0.1],
    "max_depth": [2, 3, 4],
    "subsample": [0.8, 1.0]
}

# Grid search (3-fold CV = faster)
grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=3,
    scoring="r2",
    n_jobs=-1,   # use all cores
    verbose=1
)

# Fit grid search
grid_search.fit(X_train, y_train)

print("Best parameters:", grid_search.best_params_)
print("Best CV score:", grid_search.best_score_)

# Use best model
best_rf = grid_search.best_estimator_

# Predictions
y_pred_train = best_rf.predict(X_train)
y_pred_test = best_rf.predict(X_test)

# Metrics
rf_train_r2 = r2_score(y_train, y_pred_train)
rf_test_r2 = r2_score(y_test, y_pred_test)
rf_train_mse = mean_squared_error(y_train, y_pred_train)
rf_test_mse = mean_squared_error(y_test, y_pred_test)
rf_train_mae = mean_absolute_error(y_train, y_pred_train)
rf_test_mae = mean_absolute_error(y_test, y_pred_test)

print(f"R² on training set  : {rf_train_r2:.4f}")
print(f"R² on test set      : {rf_test_r2:.4f}")
print(f"MSE on training set : {rf_train_mse:.4f}")
print(f"MSE on test set     : {rf_test_mse:.4f}")
print(f"MAE on training set : {rf_train_mae:.4f}")
print(f"MAE on test set     : {rf_test_mae:.4f}")
```
